Print-Display Result/Output of code to User

```python
print('Hello, "World"')
print("Hello \"PYTHON\"") #Escape error of double qoute
print("Hello \'PYTHON\'") #Escape error of single qoute
print("Massage1 \nMassage2") #New line
print("Massage1 \tMassage2") #tab
print("path: C:\\Users\\drive\\abc") #escape backspace add \
```

```python
name = "Atul"
print("Hello My Name is",name) # coma will add automatic space
print(name,"is learning Python")
print(name,"Wants to become a Python expert")
```
